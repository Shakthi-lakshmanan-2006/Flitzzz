```python
import pandas as pd
import numpy as np
import os

file_path = r"C:\Users\Shakthi\Desktop\Flitzz\data\flights.csv"

flights = pd.read_csv(
    file_path,
    dtype={
        "AIRLINE": "string",
        "TAIL_NUMBER": "string",
        "ORIGIN_AIRPORT": "string",
        "DESTINATION_AIRPORT": "string",
        "CANCELLATION_REASON": "string"
    }
)

print("Loaded successfully")
print("Rows:", len(flights))
print("Columns:", len(flights.columns))
```

```python
# Remove unwanted columns
columns_to_remove = [
    "CANCELLATION_REASON",
    "AIR_SYSTEM_DELAY",
    "SECURITY_DELAY",
    "AIRLINE_DELAY",
    "LATE_AIRCRAFT_DELAY",
    "WEATHER_DELAY"
]

flights.drop(columns=columns_to_remove, inplace=True)

# Remove rows where target/outcome is unavailable
flights.dropna(
    subset=["ARRIVAL_DELAY", "SCHEDULED_TIME"],
    inplace=True
)

# Remove cancelled and diverted flights
flights = flights[
    (flights["CANCELLED"] == 0) &
    (flights["DIVERTED"] == 0)
].copy()

# Target
flights["DELAYED_15"] = (
    flights["ARRIVAL_DELAY"] > 15
).astype("int8")

# Time features
flights["DEPARTURE_HOUR"] = (
    flights["SCHEDULED_DEPARTURE"] // 100
).astype("int8")

flights["DEPARTURE_MINUTE"] = (
    flights["SCHEDULED_DEPARTURE"] % 100
).astype("int8")

flights["ARRIVAL_HOUR"] = (
    flights["SCHEDULED_ARRIVAL"] // 100
).astype("int8")

flights["ARRIVAL_MINUTE"] = (
    flights["SCHEDULED_ARRIVAL"] % 100
).astype("int8")

# Weekend
flights["IS_WEEKEND"] = (
    flights["DAY_OF_WEEK"] >= 6
).astype("int8")

# Season
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

flights["SEASON"] = flights["MONTH"].apply(get_season)

# Departure period
def get_departure_period(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

flights["DEPARTURE_PERIOD"] = (
    flights["DEPARTURE_HOUR"]
    .apply(get_departure_period)
)

print("Final shape:", flights.shape)
```

```python
os.makedirs("checkpoints", exist_ok=True)

checkpoint_path = os.path.abspath(
    "checkpoints/flights_cleaned_time_features.csv"
)

flights.to_csv(
    checkpoint_path,
    index=False
)

print("CHECKPOINT SAVED!")
print(checkpoint_path)
```

```python
import pandas as pd
import numpy as np

checkpoint_path = r"C:\Users\Shakthi\checkpoints\flights_cleaned_time_features.csv"

flights = pd.read_csv(
    checkpoint_path,
    dtype={
        "AIRLINE": "string",
        "TAIL_NUMBER": "string",
        "ORIGIN_AIRPORT": "string",
        "DESTINATION_AIRPORT": "string",
        "SEASON": "string",
        "DEPARTURE_PERIOD": "string"
    }
)

print("Loaded successfully!")
print("Shape:", flights.shape)
```

```python
flights["ROUTE"] = (
    flights["ORIGIN_AIRPORT"].astype("string")
    + "_"
    + flights["DESTINATION_AIRPORT"].astype("string")
)

print("ROUTE created successfully!")
print("Unique routes:", flights["ROUTE"].nunique())
```

```python
flights[
    [
        "ORIGIN_AIRPORT",
        "DESTINATION_AIRPORT",
        "ROUTE"
    ]
].head(10)
```

```python
flights["FLIGHT_DATETIME"] = (
    pd.to_datetime(
        flights[["YEAR", "MONTH", "DAY"]]
    )
    + pd.to_timedelta(
        flights["SCHEDULED_DEPARTURE"] // 100,
        unit="h"
    )
    + pd.to_timedelta(
        flights["SCHEDULED_DEPARTURE"] % 100,
        unit="m"
    )
)
```

```python
flights = flights.sort_values(
    ["FLIGHT_DATETIME", "FLIGHT_NUMBER"]
).reset_index(drop=True)

print("Flights sorted chronologically.")
```

```python
flights.head()
```

```python
flights["ROUTE_PREVIOUS_FLIGHTS"] = (
    flights.groupby(
        "ROUTE",
        sort=False
    ).cumcount()
)
```

```python
route_cumulative_delays = (
    flights.groupby(
        "ROUTE",
        sort=False
    )["DELAYED_15"]
    .cumsum()
)

flights["ROUTE_PREVIOUS_DELAYS"] = (
    route_cumulative_delays
    - flights["DELAYED_15"]
)
```

```python
flights["ROUTE_DELAY_RATE"] = np.where(
    flights["ROUTE_PREVIOUS_FLIGHTS"] > 0,
    flights["ROUTE_PREVIOUS_DELAYS"]
    / flights["ROUTE_PREVIOUS_FLIGHTS"],
    np.nan
)
```

```python
# Number of previous flights by airline
flights["AIRLINE_PREVIOUS_FLIGHTS"] = (
    flights.groupby(
        "AIRLINE",
        sort=False
    ).cumcount()
)

# Cumulative delays by airline
airline_cumulative_delays = (
    flights.groupby(
        "AIRLINE",
        sort=False
    )["DELAYED_15"]
    .cumsum()
)

# Remove current flight's outcome
flights["AIRLINE_PREVIOUS_DELAYS"] = (
    airline_cumulative_delays
    - flights["DELAYED_15"]
)

# Historical airline delay rate
flights["AIRLINE_DELAY_RATE"] = np.where(
    flights["AIRLINE_PREVIOUS_FLIGHTS"] > 0,
    flights["AIRLINE_PREVIOUS_DELAYS"]
    / flights["AIRLINE_PREVIOUS_FLIGHTS"],
    np.nan
)
```
